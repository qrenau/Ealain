```python
import os
import array
import math
import numpy as np
import pandas as pd

import shutil
```

```python
# Remove timeout and restart algo

raw_algo = pd.read_csv("raw_algo.csv",sep=",")

index_to_drop = []
raw_algo.iloc[0]
for i in range(len(raw_algo)):
    if raw_algo.iloc[i]["log"] == "timeout":
        index_to_drop.append(i)
    elif "restart" in raw_algo.iloc[i]["algo"]:
        index_to_drop.append(i)

        

d = raw_algo.copy()
d = d.drop(d.index[index_to_drop])

d.to_csv("raw_algo_parsed.csv",index = False)
```

```python
# Remove instances with one solver only

raw_algo_parsed = pd.read_csv("raw_algo_parsed.csv",sep=",")

instance_to_drop = []
iid = raw_algo_parsed["id"].unique()
for i in iid:
    instances = raw_algo_parsed[raw_algo_parsed["id"] == i]
    nb_algos = instances["algo"].unique()
    if len(nb_algos) != 2:
        instance_to_drop.append(i)

index_to_drop = []
for i in range(len(raw_algo_parsed)):
    if raw_algo_parsed.iloc[i]["id"] in instance_to_drop:
        index_to_drop.append(i)
        
d = raw_algo_parsed.copy()
d = d.drop(d.index[index_to_drop])

d.to_csv("raw_algo_parsed2.csv",index = False)
```

```python
# Remove instances with less than 10 runs

raw_algo_parsed = pd.read_csv("raw_algo_parsed2.csv",sep=",")

instance_to_drop = []
iid = raw_algo_parsed["id"].unique()
for i in iid:
    instances = raw_algo_parsed[raw_algo_parsed["id"] == i]
    nb_runs = instances["rid"].unique()
    if len(nb_runs) != 10:
        instance_to_drop.append(i)

print(instance_to_drop)
index_to_drop = []
for i in range(len(raw_algo_parsed)):
    if raw_algo_parsed.iloc[i]["id"] in instance_to_drop:
        index_to_drop.append(i)
   


d = raw_algo_parsed.copy()
d = d.drop(d.index[index_to_drop])

d.to_csv("raw_algo_parsed3.csv",index = False)
```

```python
algo_perf = pd.read_csv("algo_perf.csv",sep=",")


iid = list(algo_perf["id"].unique())
algo = algo_perf["algo"].unique()

best_algo = []
best_perf = []
median_eax = []
median_lkh = []

instances = algo_perf[algo_perf["id"] == "national_mu1979"]
df_algo = instances[instances["algo"] == "eax"]
np.median(df_algo["perf"])

for i in iid:
    perf_lkh = []
    perf_eax = []
    instances = algo_perf[algo_perf["id"] == i]
    for a in algo:
        df_algo = instances[instances["algo"] == a]
        perf = np.median(df_algo["perf"])
        if a == "lkh":
            median_lkh.append(perf)
        else:
            median_eax.append(perf)

for perf in range(len(median_eax)):
    if median_eax[perf] < median_lkh[perf]:
        best_algo.append("eax")
        best_perf.append(median_eax[perf])
    else:
        best_algo.append("lkh")
        best_perf.append(median_lkh[perf])
        
dico = {"iid": iid, "best_algo": best_algo, "best_perf": best_perf, "median_eax": median_eax, "median_lkh": median_lkh}

d = pd.DataFrame(dico)
d.to_csv("best_algo.csv", index = False)
```
