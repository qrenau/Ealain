```python
import os
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("instance_info.csv",header=None)

size = data.iloc[0,0]-1
nb_cameras = data.iloc[0,2]
nb_constraint = data.iloc[0,3]
nb_vertex = data.iloc[0,4]

x_constraint = []
y_constraint = []

constraints = []

for i in range(nb_constraint*nb_vertex*2):
    if i % 2 == 0:
        x_constraint.append(data.iloc[0,5+i])
    else:
        y_constraint.append(data.iloc[0,5+i])

    
x_cam = []
y_cam = []
start_index = nb_constraint*nb_vertex*2+5
for coord in range(nb_cameras*2):
    if coord % 2 == 0:
        x_cam.append(data.iloc[0,coord+start_index])
    else:
        y_cam.append(data.iloc[0,coord+start_index])
        
        
#### PLOT

# Square
plt.plot([0,size,size,0,0],[0,0,size,size,0], c= "black")

# Cameras
plt.scatter(x_cam,y_cam, marker= "x", c = "orange")

# Constraints
for i in range(nb_constraint):
    x = x_constraint[i*nb_vertex:(i+1)*nb_vertex] + [x_constraint[i*nb_vertex]]
    y = y_constraint[i*nb_vertex:(i+1)*nb_vertex] + [y_constraint[i*nb_vertex]]
    plt.plot(x,y, marker= "o", c = "blue",alpha =0.1)


plt.show()
```
